```python
%%R
# R code runs here using the %%R cell magic

# Create a simple data frame
df_r <- data.frame(
  name    = c("Alice", "Bob", "Charlie", "David"),
  age     = c(25, 32, 37, 29),
  salary  = c(50000, 75000, 90000, 62000)
)

# Filter high earners
high_earners <- df_r[df_r$salary > 60000, ]
print(high_earners)

# Add a new column
df_r$salary_after_raise <- df_r$salary * 1.10
print(df_r)

# Group summary using aggregate
summary_r <- aggregate(salary ~ name, data = df_r, FUN = mean)
print(summary_r)
```

```python
from pathlib import Path

repo_root = Path(r"C:\Language_Interfacing")
gitignore_path = repo_root / ".gitignore"
ignore_entries = ["data/*", "output/*"]

current_text = gitignore_path.read_text(encoding="utf-8") if gitignore_path.exists() else ""
existing_lines = {line.strip() for line in current_text.splitlines()}

added = []
for entry in ignore_entries:
    if entry not in existing_lines:
        current_text += ("" if (not current_text or current_text.endswith("\n")) else "\n") + entry + "\n"
        existing_lines.add(entry)
        added.append(entry)

if added:
    gitignore_path.write_text(current_text, encoding="utf-8")
    print("Added to .gitignore:")
    for entry in added:
        print(f"- {entry}")
else:
    print(".gitignore already contains data/* and output/*")
```

```python
from git import Repo
from pathlib import Path

repo_root = Path(r"C:\Language_Interfacing")
repo = Repo(repo_root)

# Stage all changes under code/ and code/input_code/
repo.git.add("--", "code", "code/input_code")

staged = repo.git.diff("--cached", "--name-only").strip().splitlines()
if staged:
    print("Staged files:")
    for file_name in staged:
        print(f"- {file_name}")
    print("\nChanges are staged and ready to be committed.")
else:
    print("No changes found under code/ or code/input_code/ to stage.")
```

```python
from git import Repo
from pathlib import Path

repo_root = Path(r"C:\Language_Interfacing")
repo = Repo(repo_root)

staged = [name for name in repo.git.diff("--cached", "--name-only").splitlines() if name.strip()]
if staged:
    commit_message = "Commit staged code changes"
    commit = repo.index.commit(commit_message)
    print(f"Committed: {commit.hexsha[:7]} - {commit_message}")

    origin = repo.remote(name="origin")
    push_info = origin.push("main")[0]
    print(f"Pushed to origin/main: {push_info.summary}")
else:
    print("No staged changes to commit or push.")
```
