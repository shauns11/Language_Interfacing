```python
#creating a CSV file (df) using pandas
import pandas as pd
import os

# Create a small dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [24, 27, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}

# Convert the dictionary to a pandas DataFrame
df = pd.DataFrame(data)

# Define the directory and file name
directory = ("C:/Language_Interfacing/data")
file_name = ('file1.csv')
file_path = os.path.join(directory, file_name)

# Save the DataFrame to a CSV file in the specified directory
df.to_csv(file_path, index=False)

# Display the DataFrame
print(df)

```

```python
#Load this CSV file into Python:
import pandas as pd
import os

# Define the directory and file name
directory = ("C:/Language_Interfacing/data")
file_name = 'file1.csv'
file_path = os.path.join(directory, file_name)

# Import the CSV file into a DataFrame
df = pd.read_csv(file_path)

# Display the imported DataFrame
print(df)
```

```python
#Transfer df to Stata for analysis
#import it back to python after analysis in stata
import pandas as pd
import os
#import pystata

# Transfer the DataFrame to Stata
from pystata import stata
stata.pdataframe_to_data(df, force=True)

# Perform analysis in Stata
stata.run("""
          describe
          summarize 
          tab1 City
          generate age2=Age^2
          """)

# Save the dataset in Stata format
stata_file_path = os.path.join(directory, 'file2.dta')
stata.run(f'save "{stata_file_path}", replace')

# Display the Stata dataset
stata.run(f'use "{stata_file_path}", clear')
stata.run('describe')

# Import the Stata dataset back into Python
df_from_stata = stata.pdataframe_from_data()

# Display the imported DataFrame from Stata
print(df_from_stata)
```

```python
from git import Repo
from pathlib import Path

repo_root = Path(r"C:\Language_Interfacing")
repo = Repo(repo_root)

# Ensure target data files are ignored in git
gitignore_path = repo_root / ".gitignore"
ignore_entries = ["data/file2.dta", "data/file1.csv"]

current_text = gitignore_path.read_text(encoding="utf-8") if gitignore_path.exists() else ""
existing_lines = {line.strip() for line in current_text.splitlines()}

for ignore_entry in ignore_entries:
    if ignore_entry not in existing_lines:
        separator = "" if (not current_text or current_text.endswith("\n")) else "\n"
        current_text = f"{current_text}{separator}{ignore_entry}\n"
        existing_lines.add(ignore_entry)
        print(f"Added to .gitignore: {ignore_entry}")
    else:
        print(f"Already in .gitignore: {ignore_entry}")

gitignore_path.write_text(current_text, encoding="utf-8")

```

```python
repo.index.add([r".gitignore", r"code\Stata_within_Python_2.ipynb"])
print("Staged: .gitignore and code\\Stata_within_Python_2.ipynb")

if repo.is_dirty(index=True, working_tree=True, untracked_files=False):
    commit_message = "Update Stata notebook final block and gitignore"
    commit = repo.index.commit(commit_message)
    print(f"Committed: {commit.hexsha[:7]} - {commit_message}")
else:
    print("No tracked changes to commit")

origin = repo.remote(name="origin")
push_info = origin.push("main")[0]
print(f"Pushed to origin/main: {push_info.summary}")
```
